```python
# Object detection using an image
from ultralytics import YOLO

# Load pretrained YOLO model
model = YOLO("yolo11n.pt")

# Perform object detection
results = model("C:/Users/GISHA/Documents/AI/opencv/street.jpg")

# Display detection results
for result in results:
    result.show()
```

```python
# Explanation:
# - YOLO("yolo11n.pt") loads a pretrained YOLO model.
# - model("street.jpg") performs object detection.
# - result.show() displays the image with detected objects and bounding boxes.
```

```python
# Real-time object detection using webcam

from ultralytics import YOLO
import cv2

# Load YOLO model
model = YOLO("yolo26n.pt")

# Open webcam
cap = cv2.VideoCapture(0)

while True:

    ret, frame = cap.read()

    if not ret:
        break

    # Perform object detection
    results = model(frame)

    # Draw bounding boxes
    annotated_frame = results[0].plot()

    # Display output
    cv2.imshow("YOLO Object Detection", annotated_frame)

    # Press Q to exit
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
# YOLO for Real-Time Object Detection
# YOLO (You Only Look Once) is a deep learning-based computer vision algorithm used to detect and identify multiple objects in an image or video in real time.

# It is widely used in:
# - Self-driving vehicles
# - CCTV surveillance
# - Face and person detection
# - Traffic monitoring
# - Industrial quality inspection
# - Smart parking systems
# - Robotics
# - Crowd monitoring
```

```python
# Important YOLO concepts
# A. Bounding Box

# A rectangular box drawn around a detected object.

# B. Class Label

# The name of the detected object, such as person, car, dog or bicycle.

# C. Confidence Score

# A value indicating how confident the model is about its prediction.

# D. Non-Maximum Suppression (NMS)

# A technique that removes duplicate overlapping bounding boxes and keeps the most relevant detection.

# E. Object Classes

# YOLO models can recognize multiple categories of objects. A pretrained COCO model, for example, recognizes 80 object categories.
```

```python
# Detect specific objects
from ultralytics import YOLO

model = YOLO("yolo11n.pt")

results = model(
    "C:/Users/GISHA/Documents/AI/opencv/street.jpg",
    classes=[0, 2]
)

for result in results:
    result.show()


# In the COCO dataset:
# - Class 0 = Person
# - Class 2 = Car
```
